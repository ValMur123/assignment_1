```python
# 1. Creación de la tupla de edades especificada en la asignación
ages = (20, 25, 22, 30, 28)
```

```python
# 2. Imprimir la tupla completa
print("Tupla completa de edades:", ages)
```

```python
# 3. Mostrar el primer y el último elemento
print("Primer elemento:", ages[0])
print("Último elemento:", ages[-1])
```

```python
# 4. Mostrar el valor máximo, el valor mínimo y el número de elementos
print("Valor máximo:", max(ages))
print("Valor mínimo:", min(ages))
print("Número de elementos:", len(ages))
```
