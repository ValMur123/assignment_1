```python
grades = [15, 18, 12, 20, 16]
print("Original list:", grades)
```

```python
grades.append(17)
print("After append:", grades)
```

```python
grades.sort()
print("Sorted list:", grades)
```

```python
highest = max(grades)
lowest = min(grades)
total = len(grades)

print("Highest grade:", highest)
print("Lowest grade:", lowest)
print("Total:", total)
```

```python
print("Final list:", grades)
```
